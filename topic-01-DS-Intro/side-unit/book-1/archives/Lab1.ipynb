# Part 9: Data Cleaning

```python
nesarc_data['S3AQ3C1']=nesarc_data['S3AQ3C1'].replace(' ', np.nan)
```

```python
nesarc_data['S3AQ3C1']=pd.to_numeric(nesarc_data['S3AQ3C1'], errors='coerce')
```

```python
nesarc_data['S3AQ3C1'].dtype
```
